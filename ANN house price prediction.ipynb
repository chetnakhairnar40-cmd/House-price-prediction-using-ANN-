```python
#ordinal encoding
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[["Excellent","Good","Fair","Poor"]])
house["Condition"]=oe.fit_transform(house[["Condition"]])
house
                                          
```

```python
#one hot or nominal encoding
house=pd.get_dummies(house,columns=["Garage"],drop_first=True,dtype=int)
house
```

```python
#ordinal encoding
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[["Downtown", "Suburban", "Urban", "Rural"]])
house["Location"]=oe.fit_transform(house[["Location"]])
house
                
```

```python
#Separate X and Y
X=house.drop(columns="Price")
Y=house["Price"]
```

```python
print("X shape:",X.shape)
print("Y shape:",Y.shape)
```

```python
#Train-Test split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
print("X-train:",X_train.shape)
print("X-test:",X_test.shape)
print("Y-train:",Y_train.shape)
print("Y-test:",Y_test.shape)
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
y_scaler = StandardScaler()

Y_train = y_scaler.fit_transform(Y_train.values.reshape(-1,1))
Y_test = y_scaler.transform(Y_test.values.reshape(-1,1))
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
#design of ANN model
house=Sequential()

house.add(Dense(input_dim=9,units=32,activation="relu"))
house.add(Dense(units=16,activation="relu"))
house.add(Dense(units=8,activation="relu"))
house.add(Dense(1))
```

```python
#compile the data
house.compile(optimizer="adam",loss="mean_squared_error",metrics=["mae"])
```

```python
#fit the train data
house.fit(X_train,Y_train,batch_size=32,epochs=200,validation_split=0.2)
```

```python
#predict on train data
ypred_train=house.predict(X_train)
```

```python
#evaluate the train data
from sklearn.metrics import mean_squared_error
train_mse=mean_squared_error(Y_train,ypred_train)
train_rmse=np.sqrt(train_mse)
print("train RMSE:",train_rmse)
```

```python
#predict on test data
ypred_test=house.predict(X_test)
```

```python
#Evaluatethe test data
test_mse=mean_squared_error(Y_test,ypred_test)
test_rmse=np.sqrt(test_mse)
print("test RMSE:",test_rmse)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

r2 = r2_score(Y_test, ypred_test)
mae = mean_absolute_error(Y_test, ypred_test)
mse = mean_squared_error(Y_test, ypred_test)

print("R2 Score:", r2)
print("MAE:", mae)
print("MSE:", mse)
```
